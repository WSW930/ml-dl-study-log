```python
## Exercise 1 - Use torch.linspace

import torch

x = torch.linspace(0, 1, steps = 10)
print(f"x: {x}")
print(f"x + 5: {x + 5}")
```

```python
## Exercise 2 - Indexing & Slicing

import torch

x = torch.arange(1, 17)
x = x.reshape(4,4) # 1차원 x를 4×4 텐서로 변환
print(f"x: {x}")

print(f"2행 3열의 원소: {x[1,2]}") # 0행 0열부터 인덱스 시작

print(f"마지막 두 행: {x[2:, :]}")
```

```python
## Exercise 3 - Use torch.randn & Boolean Indexing

import torch

x = torch.randn(5,5) # 평균이 0이고 표준편차가 1인 표준 정규 분포 난수로 구성된 5×5 텐서

print(f"원본 텐서: {x}")

flag = x < 0

print(f"불리언 마스크: {flag}")

print(f"0보다 작은 원소들: {x[flag]}")
```

```python
## Exercise 4 - Use torch.stack() & torch.cat()

import torch

a = torch.arange(1, 4)
b = torch.arange(4, 7)

# stack은 텐서를 쌓아올린다(적층)
stacked_tensor = torch.stack([a, b])
print(f"stacked 텐서: {stacked_tensor}")
print(f"stacked 텐서의 shape: {stacked_tensor.shape}")

# cat은 텐서를 이어붙인다(연결)
cated_tensor = torch.cat([a, b], dim = 0)
print(f"cat 결과: {cated_tensor}")
```

```python
## Exercise 5 - CPU VS GPU: Compare the time to matrix-multiply

import torch
import time

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

a_cpu = torch.rand(5000,5000)
b_cpu = torch.rand(5000,5000)

start = time.time()
_ = torch.mm(a_cpu, b_cpu)
end = time.time()
print(f"CPU 시간: {end - start: .4f}sec")

a_gpu = a_cpu.to(device)
b_gpu = b_cpu.to(device)

torch.cuda.synchronize
start = time.time()
_ = torch.mm(a_gpu, b_gpu)
end = time.time()
print(f"GPU 시간: {end - start: .4f}sec")
```
