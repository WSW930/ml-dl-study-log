```python
## 순전파 예시: 피자 가격 예측 모델

def pizza_price_model(size):
  # 피자 크기 → 가격 예측
  weight = 2
  bias = 3
  price = weight * size + bias # 선형 계산 y = wx + b 형태
  return price

# 실제 사용
size = 5 # 5인치 피자
predicted_price = pizza_price_model(5)
print(f"{size}인치 피자 예상 가격: {predicted_price}천원")
```

```python
## 간단한 선형 모델의 순전파 과정

import torch
def forward_pass(x, W, b):
  '''
  순전파 함수:
  - x: 입력 텐서([N, in_dim])
  - W: 가중치 행렬([in_dim, out_dim])
  - b: 바이어스([1 out_dim])
  '''

  z = torch.mm(x, W) + b # 가중치 곱셈과 바이어스 추가
  output = torch.relu(z) # 활성화 함수 적용
  return output

x = torch.tensor([[1.0, 2.0], [3.0, 4.0]])
W = torch.tensor([[0.5, -0.5], [1.0, 1.0]])
b = torch.tensor([[1.0, 1.0]])

output = forward_pass(x, W, b)
print(f"순전파 출력: \n {output}")
```

```python
## 간단한 선형 회귀 모델

import torch
import torch.nn as nn

# 예제: y = 2x + 1 관계의 선형 회귀 모델
# 입력 데이터(x)와 실제 정답(y_true) 생성

x = torch.tensor([[1.0], [2.0], [3.0], [4.0]])
y_true = 2*x + 1

# 임의의 모델 예측값(예: 실제와 약간 차이를 지닌 예측값)
model_output = torch.tensor([[2.5], [3.5], [4.5], [5.5]])
# 평균 제곱 오차(MSE) 손실 함수 정의
criterion = nn.MSELoss()
# 손실 계산
loss = criterion(model_output, y_true)
print(f"손실값(MSE): {loss.item()}")
```

```python
## 피자 가격 예측 모델의 오차 보정

# 실제 가격은 11천원이었는데 13천원으로 예측! → 오차 2천원

actual_price = 11
predicted_price = 13
error = predicted_price - actual_price
size = 5

# 역전파: 오차를 거슬러가며 가중치 수정
print(f"오차: {error}천원 비싸게 예측")
print("가중치를 줄여야 한다!")

# 기존 모델 파라미터
weight = 2
bias = 3

# 가중치 업데이트(매우 간단화된 버전)
learning_rate = 0.01
weight -= learning_rate * error * size
bias -= learning_rate * error

print(f"새로운 가중치: {weight: .2f}, 새로운 편향: {bias: .2f}")
```

```python
## 스칼라 함수를 사용한 역전파 과정

import torch

# 1. x에 대해 자동 미분 활성화(requires_grad = True)
x = torch.tensor(2.0, requires_grad=True)

# 2. 함수 계산
y = x**2 + 3*x + 4

# 3. 역전파 실행: y를 x에 대해 미분하여 기울기 계산
y.backward()

# 4. 결과 출력: x에 대한 기울기
print(f"x에 대한 기울기: {x.grad}") # 결과 예상: tensor(7.)

```

```python
## 2차원 텐서로 각 요소에 대한 기울기 계산

import torch

# 2차원 입력 텐서 생성(requires_grad = True)
x = torch.tensor([[1.0, 2.0],[3.0, 4.0]], requires_grad = True)

# 간단한 함수: 각 요소를 제곱하여 모든 계산
y = (x ** 2).sum()

# 역전파 실행: 각전체 합에 대한 x의 기울기 계산
y.backward()

# 4. 결과 출력: 각 요소에 대한 기울기
print(f"x에 대한 기울기: {x.grad}")
# 결과 예상: 각 요소에 대한 미분 값은 2*x이므로
# tensor([[2.0, 4.0], [6.0, 8.0]])
```

```python
## 간단한 선형 회귀 모델을 통해 구현한 파라미터 업데이트 과정

import torch
import torch.nn as nn
import torch.optim as optim

# 선형 회귀 모델 정의: y = Wx + b
class LinearegressionModel(nn.Module):
  def __init__(self, input_dim, output_dim):
    super(LinearegressionModel, self).__init__()
    self.linear = nn.Linear(input_dim, output_dim)
  def forward(self, x):
    return self.linear(x)

# 예제 데이터: y = 2x + 1
x = torch.tensor([[1.0], [2.0], [3.0], [4.0]])
y_true = 2*x + 1

# 모델, 손실 함수, 옵티마이저 설정
model = LinearegressionModel(1, 1)
criterion = nn.MSELoss()
optimizer = optim.SGD(model.parameters(), lr=0.01)

# 학습 과정: 100 에포크 동안 파라미터 업데이트 수행
for epoch in range(100):
  optimizer.zero_grad # 이전 기울기 초기화
  pred = model(x) # 예측 수행
  loss = criterion(pred, y_true) # 손실 계산
  loss.backward() # 역전파
  optimizer.step() # 파라미터 업데이트(실제 업데이트가 발생하는 지점)

  if (epoch + 1) & 20 == 0:
    print(f"Epoch [{epoch+1}/100], Loss: {loss.item():.4f}")
```
