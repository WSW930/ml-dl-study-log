```python
## MNIST 이미지 전처리: 랜덤 회전과 정규화 적용

from torchvision import datasets, transforms

# 데이터 전처리: 랜덤 회전, 텐서 변환, 정규화
train_transforms = transforms.Compose(
    [transforms.RandomRotation(10),
     transforms.ToTensor(),
     transforms.Normalize((0.1307,), (0.3081,))
     # MNIST 데이터셋의 평균: 0.1307, 표준편차: 0.3081
    ]
)

train_dataset = datasets.MNIST(root='./data', train=True, download=True, transform=train_transforms)



## DataLoader를 활용한 배치 데이터 구성

from torch.utils.data import DataLoader

# 배치 사이즈 = 64, 데이터 무작위 섞기 O
train_loader = DataLoader(train_dataset, batch_size=64, shuffle=True)
# train_loader에서 데이터 세트를 순차적으로 추출하기 위한 반복자 객체
data_iter = iter(train_loader)
images, labels = next(data_iter)

print(f"배치 이미지의 shape: {images.shape}")
print(f"배치 이미지 레이블의 shape: {labels.shape}")
```
