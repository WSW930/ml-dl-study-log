```python
### Ch 2. Basic PyTorch Operations - Comprehensive Practice Code

import torch
import numpy as np
import time


## 1. 텐서(Tensor) 생성
# torch.arange → 1부터 5까지 정수를 담은 텐서 a & 6부터 10까지 정수를 담은 텐서 b 생성

a = torch.arange(1, 6)
b = torch.arange(6, 11)

print("텐서 a:", a)
print("텐서 b:", b)


## 2. 기본 연산
# 텐서 a와 b의 원소별 덧셈과 곱셈 수행

print(f"a + b = {a + b}")
print(f"a * b = {a * b}")


## 3. 브로드캐스팅
# 텐서 c는 (3,1) 형태 & 텐서 d는 (3,) 형태
# 두 텐서를 더하면 브로드캐스팅 규칙에 따라 d가 (3,1)로 확장 → (3,3) 형태 결과 생성

c = torch.tensor([[1],[2],[3]])
d = torch.tensor([10, 20, 30])
print(f"브로드캐스팅 c + d: \n{c + d}")


## 4. Numpy 비교
# 텐서 a, b를 Numpy 배열로 변환 후 동일한 연산 결과를 비교한다.

np.a = a.numpy()
np.b = b.numpy()
np_result = np.a + np.b # 넘파이 덧셈
pt_result = a + b # 파이토치 덧셈
print(f"Numpy 덧셈 결과: {np_result}")
print(f"Pytorch 덧셈 결과: {pt_result}")


## 5. GPU 성능 비교
# 대규모 행렬 곱셈을  통해 CPU와 GPU 연산 시간을 비교한다.

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
# 크기가 큰 랜덤 텐서 생성(10000 × 10000)
x_cpu = torch.rand((10000,10000))
y_cpu = torch.rand((10000,10000))

# CPU에서의 행렬 곱셈 연산 시간 측정
start = time.time()
_ = torch.mm(x_cpu, y_cpu)
cpu_time = time.time() - start
print(f"CPU matmul time: {cpu_time:.4f}sec")

# 생성한 텐서를 GPU로 이동
x_gpu = x_cpu.to(device)
y_gpu = y_cpu.to(device)

# gpu 연산은 비동기적으로 수행되므로 결과 측정을 위해 동기화
torch.cuda.synchronize()
start = time.time()
_ = torch.mm(x_gpu, y_gpu)
gpu_time = time.time() - start
print(f"GPU matmul time: {gpu_time:.4f}sec")
```
