```python
import pandas as pd
import numpy as np

perch_full = pd.read_csv('https://bit.ly/perch_csv_data')
perch_weight = np.array([5.9, 32.0, 40.0, 51.5, 70.0, 100.0, 78.0, 80.0, 85.0, 85.0, 110.0,
       115.0, 125.0, 130.0, 120.0, 120.0, 130.0, 135.0, 110.0, 130.0,
       150.0, 145.0, 150.0, 170.0, 225.0, 145.0, 188.0, 180.0, 197.0,
       218.0, 300.0, 260.0, 265.0, 250.0, 250.0, 300.0, 320.0, 514.0,
       556.0, 840.0, 685.0, 700.0, 700.0, 690.0, 900.0, 650.0, 820.0,
       850.0, 900.0, 1015.0, 820.0, 1100.0, 1000.0, 1100.0, 1000.0,
       1000.0])

from sklearn.model_selection import train_test_split
train_input, test_input, train_target, test_target = train_test_split(perch_full, perch_weight, random_state=42)

from sklearn.preprocessing import PolynomialFeatures
poly = PolynomialFeatures(include_bias = False)

poly.fit(train_input)
train_poly = poly.transform(train_input)
test_poly = poly.transform(test_input)

print(train_poly.shape)
print(poly.get_feature_names_out())

import sklearn.linear_model
lr = sklearn.linear_model.LinearRegression()
lr.fit(train_poly, train_target)
print("다중회귀 훈련 데이터 정확도:", lr.score(train_poly, train_target))
print("다중회귀 테스트 데이터 정확도:", lr.score(test_poly, test_target))

poly = PolynomialFeatures(degree = 5, include_bias = False)
poly.fit(train_input)
train_poly = poly.transform(train_input)
test_poly = poly.transform(test_input)

lr.fit(train_poly, train_target)
print("특성을 5차항까지 생성한 후 다중회귀 훈련 데이터 정확도:", lr.score(train_poly, train_target))
print("특성을 5차항까지 생성한 후 다중회귀 테스트 데이터 정확도:", lr.score(test_poly, test_target))

from sklearn.preprocessing import StandardScaler
ss = StandardScaler()
ss.fit(train_poly)
train_scaled = ss.transform(train_poly)
test_scaled = ss.transform(test_poly)

from sklearn.linear_model import Ridge
ridge = Ridge()
ridge.fit(train_scaled, train_target)

print("릿지 모델 적용 후 훈련 데이터 정확도:", ridge.score(train_scaled, train_target))
print("릿지 모델 적용 후 테스트 데이터 정확도:", ridge.score(test_scaled, test_target))

alpha_list = [0.001, 0.01, 0.1, 1, 10, 100]
train_score_list =[]
test_score_list = []

for alpha in alpha_list:
    ridge = Ridge(alpha = alpha)
    ridge.fit(train_scaled, train_target)
    train_score_list.append(ridge.score(train_scaled, train_target))
    test_score_list.append(ridge.score(test_scaled, test_target)) # 여기서 _scaled 대신 _poly를 투입하면 표준화의 중요성을 인지할 수 있다.

import matplotlib.pyplot as plt
plt.plot(alpha_list, train_score_list)
plt.plot(alpha_list, test_score_list)
plt.xscale('log')
plt.xlabel('alpha')
plt.ylabel('R^2')
plt.show()

ridge = Ridge(alpha = 0.1)
ridge.fit(train_scaled, train_target)
print("최적 alpha에 의한 릿지 훈련 데이터 정확도:", ridge.score(train_scaled, train_target))
print("최적 alpha에 의한 릿지 테스트 데이터 정확도:", ridge.score(test_scaled, test_target))

from sklearn.linear_model import Lasso
lasso = Lasso()
lasso.fit(train_scaled, train_target)

print("릿지 모델 적용 후 훈련 데이터 정확도:", lasso.score(train_scaled, train_target))
print("릿지 모델 적용 후 테스트 데이터 정확도:", lasso.score(test_scaled, test_target))

alpha_list = [0.001, 0.01, 0.1, 1, 10, 100]
train_score_list =[]
test_score_list = []

for alpha in alpha_list:
    lasso = Lasso(alpha = alpha)
    lasso.fit(train_scaled, train_target)
    train_score_list.append(lasso.score(train_scaled, train_target))
    test_score_list.append(lasso.score(test_scaled, test_target))

import matplotlib.pyplot as plt
plt.plot(alpha_list, train_score_list)
plt.plot(alpha_list, test_score_list)
plt.xscale('log')
plt.xlabel('alpha')
plt.ylabel('R^2')
plt.show()

lasso = Lasso(alpha = 10)
lasso.fit(train_scaled, train_target)
print("최적 alpha에 의한 라쏘 훈련 데이터 정확도:", lasso.score(train_scaled, train_target))
print("최적 alpha에 의한 라쏘 테스트 데이터 정확도:", lasso.score(test_scaled, test_target))
```
