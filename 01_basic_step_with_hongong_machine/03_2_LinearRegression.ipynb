```python
import numpy as np

# 입력 특성(길이)와 타깃 특성(무게) 데이터를 Numpy 배열로 저장
perch_length = np.array([8.4, 13.7, 15.0, 16.2, 17.4, 18.0, 18.7, 19.0, 19.6, 20.0, 21.0,
       21.0, 21.0, 21.3, 22.0, 22.0, 22.0, 22.0, 22.0, 22.5, 22.5, 22.7,
       23.0, 23.5, 24.0, 24.0, 24.6, 25.0, 25.6, 26.5, 27.3, 27.5, 27.5,
       27.5, 28.0, 28.7, 30.0, 32.8, 34.5, 35.0, 36.5, 36.0, 37.0, 37.0,
       39.0, 39.0, 39.0, 40.0, 40.0, 40.0, 40.0, 42.0, 43.0, 43.0, 43.5,
       44.0])
perch_weight = np.array([5.9, 32.0, 40.0, 51.5, 70.0, 100.0, 78.0, 80.0, 85.0, 85.0, 110.0,
       115.0, 125.0, 130.0, 120.0, 120.0, 130.0, 135.0, 110.0, 130.0,
       150.0, 145.0, 150.0, 170.0, 225.0, 145.0, 188.0, 180.0, 197.0,
       218.0, 300.0, 260.0, 265.0, 250.0, 250.0, 300.0, 320.0, 514.0,
       556.0, 840.0, 685.0, 700.0, 700.0, 690.0, 900.0, 650.0, 820.0,
       850.0, 900.0, 1015.0, 820.0, 1100.0, 1000.0, 1100.0, 1000.0,
       1000.0])


# train_test_split 메서드로 훈련-테스트 데이터(입력/타깃) 추출
from sklearn.model_selection import train_test_split

train_input, test_input, train_target, test_target = train_test_split(perch_length, perch_weight, random_state=42)

train_input = train_input.reshape(-1, 1)
test_input = test_input.reshape(-1, 1)

# K-NN 회귀 알고리즘 모델 생성
from sklearn.neighbors import KNeighborsRegressor

knr = KNeighborsRegressor(n_neighbors = 3)
knr.fit(train_input, train_target)
# K-NN 회귀 알고리즘의 한계 - 기존 입력 데이터를 벗어난 데이터 예측이 어렵다(아래 2줄 예시)
print("50cm 입력에 대한 K-NN 예측값:", knr.predict([[50]]))
print("100cm 입력에 대한 K-NN 예측값:", knr.predict([[100]]))


import matplotlib.pyplot as plt
# K-NN 회귀 알고리즘의 한계를 그래프에서 확인 - 예측 가능값 범위가 한정됨.
indexes = knr.kneighbors([[50]], return_distance=False)
plt.scatter(train_input, train_target)
plt.scatter(train_input[indexes], train_target[indexes], marker = 'D')
plt.xlabel('length')
plt.ylabel('weight')
plt.show()

print("50cm 입력에 대한 K-NN 예측값: , ", knr.predict([[50]]), "50cm 입력 세 이웃 타깃의 평균: ", np.mean(train_target[indexes]))


#선형 회귀 모델 생성
from sklearn.linear_model import LinearRegression

lr = LinearRegression()
lr.fit(train_input, train_target)

print("1차 선형 회귀 예측값: ",lr.predict([[50]]))
# coef_ 및 intercept_ 속성으로 모델이 도출한 파라미터(가중치/절편) 확인 가능
print("1차 선형 회귀 가중치: ", lr.coef_, lr.intercept_)

# 선형 회귀 예측을 그래프에서 확인 => 정확도 향상, 그러나 여전히 실제와 괴리 존재
plt.scatter(train_input, train_target)
plt.plot([15, 50], [15*lr.coef_+lr.intercept_, 50*lr.coef_+lr.intercept_])
plt.scatter(50, 1241.8, marker = '^')
plt.xlabel('length')
plt.ylabel('weight')
plt.show()

print(lr.score(train_input, train_target))
print(lr.score(test_input, test_target))


# 입력에 대한 2차 선형 회귀 모델 생성
train_poly = np.column_stack((train_input**2, train_input))
test_poly = np.column_stack((test_input**2, test_input))
print(train_poly.shape, test_poly.shape)

lr = LinearRegression()
lr.fit(train_poly, train_target)

print("2차 선형 회귀 예측값:", lr.predict([[50**2, 50]]))
print("2차 선형 회귀 가중치: ", lr.coef_, lr.intercept_)

# 2차 선형 회귀의 입력으로 활용할 자연수 배열(15~50) 생성
point = np.arange(15, 50)

plt.scatter(train_input, train_target)
# 2차 선형 회귀, 여러 개의 가중치
plt.plot(point, 1.01*point**2 - 21.6 * point + 116.05)

plt.scatter(50, 1574, marker = '^')
plt.xlabel('length')
plt.ylabel('weight')
plt.show()
```
