```python
# pandas 라이브러리로 외부 데이터 호출 & 저장
import pandas as pd

fish = pd.read_csv('https://bit.ly/fish_csv_data')
fish_input = fish[['Weight', 'Length', 'Diagonal', 'Height', 'Width']]
fish_target = fish['Species']



# 훈련/테스트 입력-타깃 추출
from sklearn.model_selection import train_test_split

train_input, test_input, train_target, test_target = train_test_split(fish_input, fish_target, random_state=42)

# 데이터 전처리 - 표준화 작업(특성 스케일 차이 조정)
from sklearn.preprocessing import StandardScaler

ss = StandardScaler()
ss.fit(train_input)
train_scaled = ss.transform(train_input)
test_scaled = ss.transform(test_input)



# 확률적 경사 하강법 객체 생성
from sklearn.linear_model import SGDClassifier

# loss 속성을 log_loss(로지스틱 손실함수)로 설정, 반복 10회
sc = SGDClassifier(loss = 'log_loss', max_iter = 10, random_state = 42)
sc.fit(train_scaled, train_target)
# 경사하강법 - 최초 훈련만으로는 정확도 부족(과소적합)
print(sc.score(train_scaled, train_target))
print(sc.score(test_scaled, test_target))

# partial_fit() 메서드 => 경사하강법 객체의 점진적 추가 훈련 => 에포크 반복 시 훈련 정확도 향상
sc.partial_fit(train_scaled, train_target)
print(sc.score(train_scaled, train_target))
print(sc.score(test_scaled, test_target))




# 에포크 반복에 따른 과소적합-적정-과대적합 파악
import numpy as np
sc = SGDClassifier(loss = 'log_loss', random_state = 42)

train_score = []
test_score = []
classes = np.unique(train_target)

# 에포크 300회 반복문
for _ in range(0, 300):
  # ** partial_fit으로 초기 에포크(훈련) 시 classes로 클래스 집합을 입력해야 함!
  sc.partial_fit(train_scaled, train_target, classes = classes)
  train_score.append(sc.score(train_scaled, train_target))
  test_score.append(sc.score(test_scaled, test_target))



# 에포크 300회 반복에 따른 훈련-테스트 정확도 추이 그래프
import matplotlib.pyplot as plt

plt.plot(train_score)
plt.plot(test_score)
plt.xlabel('epoch')
plt.ylabel('accuracy')
plt.show()



# 최적 에포크 횟수를 수행할 때 확률적 경사 하강법 객체의 정확도
sc = SGDClassifier(loss = 'log_loss', max_iter = 100, tol = None, random_state = 42)
sc.fit(train_scaled, train_target)
print(sc.score(train_scaled, train_target))
print(sc.score(test_scaled, test_target))
```

```python
import numpy as np

# 독립 변수 x에 대한 f(x)와 df(x) 정의
def f(x):
    return x**2

def df(x):
    return 2*x




# 초기 x값, 학습률, 반복 횟수 설정
x = -10.0
learning_rate = 0.1
max_iter = 20

x_history = [x]
f_history = [f(x)]

for i in range(max_iter):
    gradient = df(x)

    # 경사하강법 수식에 따라 x 변화
    x = x - learning_rate * gradient

    # 반복 횟수마다 x, f(x) 기록, print를 통해 x와 기울기 값 출력
    x_history.append(x)
    f_history.append(f(x))
    print(f"실행 횟수: {i+1}, x값: {np.round(x, decimals = 4)}, 기울기 값: {np.round(df(x), decimals = 4)}")

import matplotlib.pyplot as plt

# matplolib 라이브러리를 통해 그래프 시각화
plt.plot(x_history, f_history, label='f(x)') # f(x) 값 추가
plt.scatter(x_history, f_history, color='red', label='Descent Steps') # 하강 단계 점 추가
plt.title('Gradient Descent Visualization')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.legend()
plt.grid(True)
plt.show()
```
