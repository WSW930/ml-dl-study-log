```python
### 인공 신경망

## 케라스 기반 패션 MNIST 데이터 호출

import keras

# keras.datasets.fashion_mnist 모듈 아래 load_data 메서드로 데이터 호출
(train_input, train_target), (test_input, test_target) =\
keras.datasets.fashion_mnist.load_data()

print(train_input.shape, train_target.shape)
print(test_input.shape, test_target.shape)



## 맷플롯립으로 패션 MNIST 데이터 시각화(1~10번째 샘플)

import matplotlib.pyplot as plt

fig, axs = plt.subplots(1, 10, figsize=(10, 10))

for i in range(10):
  axs[i].imshow(train_input[i], cmap = 'gray_r')
  axs[i].axis('off')

plt.show()

print(train_target[:10])

import numpy as np

print("MINST 카테고리 정보: ", np.unique(train_target, return_counts = True))



## 로지스틱 회귀로 패션 아이템 분류

# 방대한 데이터 => 정규 방정식보다 경사 하강법 모델이 유리(로지스틱 손실 기반)
# 경사 하강법 => "특성 스케일"이 성능에 영향을 미침 => 정규화 필요

train_scaled = (train_input / 255) # 양수값 이미지 전처리 기법(0~1 사이 정규화)
train_scaled = train_scaled.reshape(-1, 28 * 28) # 사이킷런 입력을 위한 2차원 변환

from sklearn.model_selection import cross_validate
from sklearn.linear_model import SGDClassifier

sc = SGDClassifier(loss = 'log_loss', max_iter = 5, random_state = 42)

scores = cross_validate(sc, train_scaled, train_target, n_jobs = -1)
print("로지스틱 손실 기반 SGC의 교차 검증 평균 정확도: ", np.mean(scores['test_score']))



## 인공 신경망 - 텐서플로와 케라스

import tensorflow as tf
import keras

## 인공 신경망 - 모델 만들기: 입력층 - 밀집층 - 출력층 구조)

from sklearn.model_selection import train_test_split

# 인공 신경망은 검증 세트를 직접 떼어내어 검증하는 것이 일반적
train_scaled, val_scaled, train_target, val_target = train_test_split(train_scaled, train_target, test_size = 0.2, random_state = 42)

print("ANN 훈련 세트 크기: ", train_scaled.shape, train_target.shape)
print("ANN 검증 세트 크기: ", val_scaled.shape, val_target.shape)


# 입력층/밀집층 생성 => keras.layers 모듈 메서드 사용
inputs = keras.layers.Input(shape = (28 * 28,)) # 'I'nput 대문자 유의

dense = keras.layers.Dense(10, activation = 'softmax')

model = keras.Sequential([inputs, dense]) # 입력층-밀집층을 결합한 인공 신경망 객체



## 인공 신경말 모델으로 패션 아이템 분류

# .compile 메서드로 인공 신경망 객체의 손실함수(크로스-엔트로피) 및 측정 요소(정확도)
model.compile(loss = 'sparse_categorical_crossentropy', metrics = ['accuracy'])

print(train_target[:10])

model.fit(train_scaled, train_target, epochs = 5)

# 인공 신경망 모델 - 성능 평가 시 .evaluate 메서드 활용
model.evaluate(val_scaled, val_target)



```
