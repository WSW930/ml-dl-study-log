```python
### 비지도 학습과 군집 알고리즘

## 과일 이미지 데이터 준비 & 기초 파악
!wget https://bit.ly/fruits_300_data -O fruits_300.npy

import numpy as np
import matplotlib.pyplot as plt

fruits = np.load('fruits_300.npy') # npy 외부 파일 로드 => 저장
print("과일 npy 데이터의 크기: ", fruits.shape) # (개수,높이,너비) => 3차원 배열
print(fruits[0, 0, : ]) # 첫 번째 이미지 첫 번째 행 모두 출력

plt.imshow(fruits[0], cmap = 'gray') # .imshow => numpy 배열로 저장된 이미지 그리기 메서드
plt.show()

plt.imshow(fruits[0], cmap = 'gray_r') # gray_r => 이미지 흑백을 반전시킨다.
plt.show()

fig, axs = plt.subplots(1, 2) # .subplots() => 여러 그래프를 배열처럼 쌓는(stack) 메서드. 현재는 1행 2열 세팅
axs[0].imshow(fruits[100], cmap = 'gray_r') # 참고: fruit[100]~fruit[199] => 파인애플
axs[1].imshow(fruits[200], cmap = 'gray_r') # 참고: fruit[200]~fruit[299] => 바나나
plt.show()




## 픽셀값 분석
apple = fruits[0:100].reshape(-1, 100 * 100)
pineapple = fruits[100:200].reshape(-1, 100 * 100)
banana = fruits[200:300].reshape(-1, 100 * 100)

print(apple.shape)
print(apple.mean(axis = 1))
# axis = 0; 행 방향 계산(수직) (ex-여러 장의 사과 이미지 일부 픽셀),
# axis = 1; 열 방향 계산(수평) (ex-사과 한 장의 전체 이미지?)

# 1. 사진 단위 평균 픽셀값 비교
# hist => 히스토그램(빈도 그래프) 작성, alpha => 투명도 매개변수
plt.hist(apple.mean(axis = 1), alpha = 0.8, label = 'apple')
plt.hist(pineapple.mean(axis = 1), alpha = 0.8, label = 'pineapple')
plt.hist(banana.mean(axis = 1), alpha = 0.8, label = 'banana')
plt.legend()
plt.show() # 픽셀 수 기준 => 바나나 <-> 사과, 파인애플 분포; 사과와 파인애플 구분 어려움.


# 2. 픽셀 위치별 픽셀값 비교(사진 100장의 똑같은 픽셀 위치)
fig, axs = plt.subplots(1, 3, figsize = (20, 5)) # 1행 3열(총 3개)의 서브 그래프 제작
axs[0].bar(range(10000), apple.mean(axis = 0)) #.bar => 막대 그래프 작성 메서드
axs[1].bar(range(10000), pineapple.mean(axis = 0))

axs[2].bar(range(10000), banana.mean(axis = 0)) # 픽셀 위치별 픽셀값 => 과일 종류마다 상이하다! => 구분 가능



# 평균 이미지(대표 이미지) 그리기
apple_mean = apple.mean(axis=0).reshape(100, 100)
pineapple_mean = pineapple.mean(axis=0).reshape(100, 100)
banana_mean = banana.mean(axis=0).reshape(100, 100)

fig, axs = plt.subplots(1, 3, figsize=(20, 5))
axs[0].imshow(apple_mean, cmap = 'gray_r')
axs[1].imshow(pineapple_mean, cmap = 'gray_r')
axs[2].imshow(banana_mean, cmap = 'gray_r')
plt.show()



## 평균값과 가까운 사진 고르기(사과 기준)

abs_diff = np.abs(fruits - apple_mean)
abs_mean = np.mean(abs_diff, axis=(1, 2))
# axis = (1, 2) => 두 번째, 세 번째 차원 지정 => 첫 번째 차원(축)만 남게 된다
# 300개 과일 이미지 각각의 오차 평균(총 300개의 오차 데이터 1차원 배열)

apple_index = np.argsort(abs_mean)[:100] # 오차가 가장 작은 샘플 100개(=사과 이미지) 인덱스 추출
apple_index = apple_index.reshape(10, 10)
fig, axs = plt.subplots(10, 10, figsize=(10, 10)) # 10 × 10 서브그래프 생성
for i in range(10):
  for j in range(10):
    axs[i, j].imshow(fruits[apple_index[i, j]], cmap = 'gray_r')
    axs[i, j].axis('off') # .axis('') => 좌표축 ON/OFF

plt.show()
```
